```python
%%writefile core/preprocessor.py


class TextPreprocessor:
    def clean(self, text: str) -> str:
        if not isinstance(text, str):
            return ""
        return text.strip().lower()
```

```python
%%writefile core/validator.py

class RequestValidator:
    def is_valid(self, text: str) -> bool:
        if not text:
            return False
        return len(text.strip()) >= 3
```

```python
%%writefile core/classifier.py


from typing import Tuple


class CategoryClassifier:
    def detect_gender(self, text: str) -> str:

        male_keywords = ["for my brother", "for my boyfriend", "for my husband", "male", "man", "boy", "him", "his"]
        female_keywords = ["for my sister", "for my girlfriend", "for my wife", "female", "woman", "girl", "her", "she"]

        for kw in male_keywords:
            if kw in text:
                return "male"

        for kw in female_keywords:
            if kw in text:
                return "female"

        # Default gender if not specified
        return "female"

    def detect_style(self, text: str) -> str:

        casual_words = ["casual", "everyday", "uni", "university", "college", "day out"]
        formal_words = ["formal", "wedding", "ceremony", "reception"]
        party_words = ["party", "night out", "birthday", "club", "evening"]
        business_words = ["business", "office", "meeting", "presentation", "interview", "corporate"]

        for w in party_words:
            if w in text:
                return "party"

        for w in formal_words:
            if w in text:
                return "formal"

        for w in business_words:
            if w in text:
                return "business"

        for w in casual_words:
            if w in text:
                return "casual"

        # Default style if none detected
        return "casual"

    def classify(self, text: str) -> Tuple[str, str, str]:

        gender = self.detect_gender(text)
        style = self.detect_style(text)
        category_key = f"{gender}_{style}"
        return gender, style, category_key
```

```python
%%writefile core/logger.py


from datetime import datetime


class SystemLogger:
    def log(self, message: str) -> None:
        now = datetime.now().strftime("%H:%M:%S")
        print(f"[LOG {now}] {message}")
```

```python
%%writefile core/controller.py

from typing import Tuple, List

from core.preprocessor import TextPreprocessor
from core.validator import RequestValidator
from core.classifier import CategoryClassifier
from core.recommender import RecommendationEngine
from core.logger import SystemLogger


class AppController:
    def __init__(self, recommender: RecommendationEngine):
        self.preprocessor = TextPreprocessor()
        self.validator = RequestValidator()
        self.classifier = CategoryClassifier()
        self.recommender = recommender
        self.logger = SystemLogger()

    def handle_request(self, raw_text: str) -> Tuple[str, List[str]]:
        self.logger.log(f"Received input: {raw_text!r}")
        cleaned = self.preprocessor.clean(raw_text)
        self.logger.log(f"Cleaned input: {cleaned!r}")

        if not self.validator.is_valid(cleaned):
            self.logger.log("Validation failed.")
            return "Please enter a longer description of the outfit you want.", []

        gender, style, category_key = self.classifier.classify(cleaned)
        self.logger.log(f"Classified as: gender={gender}, style={style}, key={category_key}")

        description, images = self.recommender.recommend(category_key)
        self.logger.log(f"Generated recommendation. Images: {images}")
        return description, images
```

```python
%%writefile core/recommender.py


from typing import Dict, Any, List, Tuple
import random


class RecommendationEngine:
    def __init__(self, catalogue: Dict[str, Any]):
        self.catalogue = catalogue

    def recommend(self, category_key: str) -> Tuple[str, List[str]]:
        entry = self.catalogue.get(category_key)
        if not entry:
            return (
                "Sorry, I could not clearly understand your request. "
                "Try mentioning casual, formal, party, or business plus who it's for.",
                []
            )

        templates = entry["description_templates"]
        items = entry["items"]
        images = entry["images"]

        template = random.choice(templates)
        top = random.choice(items["top"])
        bottom = random.choice(items["bottom"])
        shoes = random.choice(items["shoes"])

        description = template.format(top=top, bottom=bottom, shoes=shoes)

        chosen_images = random.sample(images, k=min(3, len(images)))

        return description, chosen_images
```

```python
from data.catalogue_loader import get_catalogue
from core.recommender import RecommendationEngine
from core.controller import AppController

from IPython.display import Image, display
import os

# Initialize layers
catalogue_data = get_catalogue()
recommender = RecommendationEngine(catalogue_data)
controller = AppController(recommender)

def show_images(image_paths):
    """
    Tries to display each image in Colab.
    If file not found, it just skips it.
    """
    for path in image_paths:
        if os.path.exists(path):
            display(Image(filename=path))
        else:
            print(f"(Image not found: {path})")

def main_menu():
    print("=== AI Fashion Stylist (Console Version) ===")
    print("1. Get outfit recommendation")
    print("2. List available categories")
    print("3. Exit")

def list_categories():
    print("\nAvailable internal categories (gender_style):")
    for key in catalogue_data.keys():
        print(" -", key)
    print()

# Simple interactive loop
while True:
    main_menu()
    choice = input("Enter your choice (1/2/3): ").strip()

    if choice == "1":
        user_text = input("\nDescribe the outfit you want (e.g., 'female casual outfit for university' or 'party outfit for my brother'): \n> ")
        description, images = controller.handle_request(user_text)
        print("\nSuggested Outfit:\n", description)
        if images:
            print("\nSample outfit images:")
            show_images(images)
        print("\n" + "-"*50 + "\n")

    elif choice == "2":
        list_categories()

    elif choice == "3":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please select 1, 2, or 3.\n")
```
